## Classification Probability

## Simple Concept

* A classification model often gives a **probability** instead of directly giving `0` or `1`.
* Example:

  * `0.90` → 90% probability of Positive
  * `0.30` → 30% probability of Positive
* We then use a **threshold** to convert this probability into the final class.

## Important Explanation

* **When to use it** – When you want to understand how confident a classification model is before making a final prediction.
* **When not to use it** – Probability is not always a perfectly calibrated real-world percentage.
* **Why we need it** – It allows us to control how easily the model predicts the Positive class.

## Python Implementation

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

# Example predicted probabilities
probabilities = [0.90, 0.70, 0.55, 0.40, 0.20]

print(probabilities)
```

## Output

```text
[0.90, 0.70, 0.55, 0.40, 0.20]
```

## What the Example Tells

* The model is more confident that `0.90` is Positive than `0.20`.
* These probabilities can later be converted into `0` or `1` using a threshold.

## Real-World Example

* A hospital model predicts the probability that a patient has a disease.

## Business Example

* A bank predicts the probability that a customer will default on a loan.

## AI/ML Use Case

* An AI/ML Engineer can use predicted probabilities to choose an appropriate classification threshold.

## Key Points

* Classification models can produce probabilities.
* Probability and final class are not the same thing.
* Threshold converts probability into a class.

## Default Threshold

## Simple Concept

* The **default threshold** for many binary classification models is `0.5`.
* If probability is:

  * `>= 0.5` → Positive (`1`)
  * `< 0.5` → Negative (`0`)

## Important Explanation

* **When to use it** – When both types of classification errors have roughly similar importance.
* **When not to use it** – When False Positive and False Negative errors have very different costs.
* **Why we need it** – It gives us a simple starting point for converting probabilities into predictions.

## Python Implementation

```python
probabilities = [0.90, 0.70, 0.55, 0.40, 0.20]

threshold = 0.5

predictions = [1 if p >= threshold else 0 for p in probabilities]

print(predictions)
```

## Output

```text
[1, 1, 1, 0, 0]
```

## What the Example Tells

* `0.90`, `0.70`, and `0.55` are classified as `1`.
* `0.40` and `0.20` are classified as `0`.
* This happened because the threshold is `0.5`.

## Real-World Example

* If a disease probability is `0.70`, the model predicts that the patient belongs to the Positive class.

## Business Example

* If a customer has a `0.65` probability of cancelling a subscription, a threshold of `0.5` predicts cancellation.

## AI/ML Use Case

* Start with `0.5`, then evaluate whether another threshold gives better results for the actual problem.

## Key Points

* `0.5` is a common default.
* `>= 0.5` → `1`.
* `< 0.5` → `0`.
* Default does not always mean optimal.

## Decision Threshold

## Simple Concept

* A **decision threshold** is the probability value used to decide whether the prediction becomes `0` or `1`.

## Important Explanation

* **When to use it** – When you need control over how sensitive the classifier is to Positive cases.
* **When not to use it** – You should not change it randomly just to improve one metric without considering the business problem.
* **Why we need it** – Different problems have different costs for False Positive and False Negative predictions.

For example:

```text
Threshold = 0.50
Probability = 0.60
Prediction = 1
```

But:

```text
Threshold = 0.70
Probability = 0.60
Prediction = 0
```

The **same probability** can produce different predictions depending on the threshold.

## Python Implementation

```python
probability = 0.60

threshold = 0.50

prediction = 1 if probability >= threshold else 0

print(prediction)
```

## Output

```text
1
```

## What the Example Tells

* Probability = `0.60`
* Threshold = `0.50`
* Because `0.60 >= 0.50`, prediction becomes `1`.

If threshold changes to `0.70`:

```text
0.60 < 0.70
```

The prediction becomes `0`.

## Real-World Example

* Disease detection:

  * Lower threshold → easier to identify possible patients.
  * Higher threshold → require stronger evidence.

## Business Example

* Fraud detection:

  * Lower threshold can flag more transactions for investigation.
  * Higher threshold can reduce unnecessary investigations.

## AI/ML Use Case

* An AI/ML Engineer chooses the threshold based on validation data and the cost of errors.

## Key Points

* Threshold controls classification decisions.
* Changing threshold changes predictions.
* Threshold is different from model training parameters.

## Threshold Tuning

## Simple Concept

* **Threshold tuning** means testing different thresholds and choosing one based on the problem's required performance.

Example:

```text
0.30
0.40
0.50
0.60
0.70
```

Instead of automatically using only `0.50`.

## Important Explanation

* **When to use it** – When the default `0.5` threshold does not give the desired Precision, Recall, or F1 Score.
* **When not to use it** – Do not tune the threshold using the test set. Use validation data, or cross-validation, to choose it and keep the test set for final evaluation.
* **Why we need it** – The model's probability ranking may be useful even when `0.5` is not the best decision point.

## Python Implementation

```python
probabilities = [0.90, 0.70, 0.55, 0.40, 0.20]

for threshold in [0.3, 0.5, 0.7]:
    predictions = [1 if p >= threshold else 0 for p in probabilities]
    
    print("Threshold:", threshold)
    print("Predictions:", predictions)
```

## Output

```text
Threshold: 0.3
Predictions: [1, 1, 1, 1, 0]

Threshold: 0.5
Predictions: [1, 1, 1, 0, 0]

Threshold: 0.7
Predictions: [1, 1, 0, 0, 0]
```

## What the Example Tells

* Lower threshold → more Positive predictions.
* Higher threshold → fewer Positive predictions.
* Therefore, changing the threshold changes the confusion matrix and evaluation metrics.

## Real-World Example

* Cancer screening may use a lower threshold to catch more potentially positive patients.

## Business Example

* An e-commerce company may lower the threshold when it wants to identify more customers likely to stop buying.

## AI/ML Use Case

* Train the model first.
* Get validation probabilities.
* Test multiple thresholds.
* Compare Precision, Recall, and F1.
* Select a threshold according to the application's objective.

## Key Points

* Threshold tuning happens after model training.
* Test several thresholds.
* Lower threshold → more Positive predictions.
* Higher threshold → fewer Positive predictions.

## Precision vs Recall Trade-off

## Simple Concept

* **Precision** asks:

  > Of the cases predicted Positive, how many were actually Positive?

* **Recall** asks:

  > Of all the actual Positive cases, how many did we find?

Usually:

```text
Lower threshold → Recall tends to increase
                 Precision may decrease

Higher threshold → Precision tends to increase
                   Recall may decrease
```

This is a common trade-off, though the exact metric values depend on the dataset.

## Important Explanation

* **When to use it** – When you need to decide whether missing Positive cases or creating extra Positive alerts is more costly.
* **When not to use it** – Do not optimize Precision or Recall blindly without understanding the consequences of errors.
* **Why we need it** – One threshold may catch more Positive cases but also create more False Positives.

## Python Implementation

```python
from sklearn.metrics import precision_score, recall_score

y_true = [1, 1, 1, 0, 0, 0]
y_pred = [1, 1, 0, 1, 0, 0]

precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)

print("Precision:", round(precision, 2))
print("Recall:", round(recall, 2))
```

## Output

```text
Precision: 0.67
Recall: 0.67
```

## What the Example Tells

* The model correctly identified some Positive cases.
* But it also made a Positive prediction for one actual Negative case.
* Precision and Recall measure different types of performance.

## Real-World Example

* Disease screening:

  * Missing a sick patient can be very costly.
  * Therefore, high Recall may be important.

## Business Example

* Spam detection:

  * Sending a genuine email to spam can be problematic.
  * Therefore, Precision can be important.

## AI/ML Use Case

* An AI/ML Engineer compares Precision and Recall at different thresholds instead of assuming `0.5` is always suitable.

## Key Points

* Precision focuses on predicted Positives.
* Recall focuses on actual Positives.
* Threshold changes can create a Precision–Recall trade-off.

## False Positive

## Simple Concept

* **False Positive (FP)** means:

  * Actual = `0`
  * Model predicted = `1`

The model says **Positive**, but the real answer is **Negative**.

## Important Explanation

* **When to use it** – Understand FP when Positive predictions can create unnecessary actions.
* **When not to use it** – Do not treat every FP as equally serious; the business impact depends on the application.
* **Why we need it** – It helps us understand the cost of incorrectly raising a Positive prediction.

## Python Implementation

```python
from sklearn.metrics import confusion_matrix

y_true = [1, 0, 1, 0]
y_pred = [1, 1, 1, 0]

cm = confusion_matrix(y_true, y_pred)

print(cm)
```

## Output

```text
[[1 1]
 [0 2]]
```

## What the Example Tells

The confusion matrix is:

```text
[[TN FP]
 [FN TP]]
```

So:

```text
TN = 1
FP = 1
FN = 0
TP = 2
```

There is **1 False Positive**.

## Real-World Example

* A healthy patient is incorrectly flagged as possibly having a disease.

## Business Example

* A genuine bank transaction is incorrectly flagged as fraudulent.

## AI/ML Use Case

* Reduce FP when unnecessary alerts, investigations, or customer friction are expensive.

## Key Points

* FP = predicted Positive, actually Negative.
* FP can reduce Precision.
* Lowering the threshold can increase FP.

## False Negative

## Simple Concept

* **False Negative (FN)** means:

  * Actual = `1`
  * Model predicted = `0`

The model says **Negative**, but the real answer is **Positive**.

## Important Explanation

* **When to use it** – Understand FN when missing a Positive case can have serious consequences.
* **When not to use it** – Do not automatically minimize FN in every problem; sometimes False Positives are more costly.
* **Why we need it** – It helps us understand cases the model failed to detect.

## Python Implementation

```python
from sklearn.metrics import confusion_matrix

y_true = [1, 0, 1, 0]
y_pred = [0, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

print(cm)
```

## Output

```text
[[2 0]
 [1 1]]
```

## What the Example Tells

```text
TN = 2
FP = 0
FN = 1
TP = 1
```

There is **1 False Negative**.

The model missed one actual Positive case.

## Real-World Example

* A patient actually has a disease, but the model predicts that the patient does not.

## Business Example

* A bank customer who will actually default is predicted as a safe customer.

## AI/ML Use Case

* Reduce FN when missing a Positive case is more costly than creating additional alerts.

## Key Points

* FN = predicted Negative, actually Positive.
* FN reduces Recall.
* Lowering the threshold can reduce FN.

## Business Cost of Errors

## Simple Concept

* Not all errors have the same cost.
* A False Positive may cost one amount.
* A False Negative may cost another amount.

Example:

```text
False Positive cost = ₹100
False Negative cost = ₹1,000
```

If FN is much more expensive, the business may prefer a threshold that catches more Positive cases, even if it creates more False Positives.

## Important Explanation

* **When to use it** – When different prediction errors have different real-world consequences.
* **When not to use it** – Do not choose a threshold only from the cheapest error if other important business, safety, or regulatory consequences exist.
* **Why we need it** – Accuracy alone does not tell us whether the model is making the *costly* mistakes.

## Python Implementation

```python
from sklearn.metrics import confusion_matrix

y_true = [1, 1, 0, 0, 1, 0]
y_pred = [1, 0, 1, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

TN, FP, FN, TP = cm.ravel()

fp_cost = 100
fn_cost = 1000

total_cost = (FP * fp_cost) + (FN * fn_cost)

print("TN:", TN)
print("FP:", FP)
print("FN:", FN)
print("TP:", TP)
print("Total business cost:", total_cost)
```

## Output

```text
TN: 2
FP: 1
FN: 1
TP: 2
Total business cost: 1100
```

## What the Example Tells

The model made:

```text
1 False Positive × ₹100  = ₹100
1 False Negative × ₹1000 = ₹1000
```

Therefore:

```text
Total cost = ₹1100
```

Even though there is only one FP and one FN, the FN contributes much more to the total cost.

## Real-World Example

* In disease screening, missing a genuinely sick patient may have a much larger consequence than sending a healthy patient for additional testing.

## Business Example

* In fraud detection:

  * FP → legitimate customer transaction is investigated.
  * FN → fraudulent transaction is allowed.
  * The business can assign different costs to these errors.

## AI/ML Use Case

An AI/ML Engineer can compare thresholds using:

```text
Threshold
    ↓
Predictions
    ↓
Confusion Matrix
    ↓
Precision / Recall / F1
    ↓
Business Cost
```

The threshold should be selected using the application's documented objective and validation data, rather than simply assuming `0.5`.

## Key Points

* Errors have different costs.
* FP and FN should be considered separately.
* Accuracy alone may hide costly errors.
* Business cost can help define the appropriate threshold.

# Overall Threshold Optimization Flow

```text
Train Classification Model
          ↓
Get Predicted Probabilities
          ↓
Choose Candidate Thresholds
          ↓
0.30 → 0.40 → 0.50 → 0.60 → 0.70
          ↓
Convert Probability → Class
          ↓
Confusion Matrix
          ↓
Precision / Recall / F1
          ↓
Consider Business Cost
          ↓
Choose Threshold Based on the Problem
          ↓
Final Evaluation on Test Data
```

## Final Revision

```text
Probability
    ↓
Threshold
    ↓
Prediction 0 or 1
    ↓
Confusion Matrix
    ↓
TP / TN / FP / FN
    ↓
Precision / Recall / F1
    ↓
Business Cost
```

### Most Important Relationship

```text
Lower Threshold
→ More Positive predictions
→ Usually higher Recall
→ Usually lower Precision
→ More FP
→ Fewer FN

Higher Threshold
→ Fewer Positive predictions
→ Usually higher Precision
→ Usually lower Recall
→ Fewer FP
→ More FN
```

The exact metric changes depend on the dataset and model; these are the usual directional effects.
